## ED_Fig. 1|Spatial distribution trends of all establishments (2015-2023)

```python
# Load data
Dback = pd.read_csv(data_path+'RT_CityD_Bldback.csv')
LayerT = pd.read_csv(data_path+'RT_CityLayertype.csv')

# Save figure
def save_figure(filename):
    plt.tight_layout()
    plt.savefig(f"{figure_path}/{filename}", format="svg", bbox_inches='tight')
    plt.show()
    
# Global settings
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 12
plt.rcParams['axes.titlesize'] = 12

# Color definitions
year_colors = {2015: '#6eaed1', 2019: '#fed06e', 2023: '#dc6e57'}
stacked_colors = ['#3851a1', '#5c90c2', '#92c6dc', '#ffe99d', '#feb76d', '#f67949', '#da382b']

# ED_Fig.1a. Distance to Street Facade
def create_EDfig1a(data):
    pivot_df = data.pivot_table(index='D_Bldback(m)', columns='Year', values='Percentage(%)', fill_value=0)
    order = ['<20', '20-40', '40-60', '60-80', '80-100', '100-120', '120-140', '>140']
    pivot_df = pivot_df.loc[order]

    plt.figure(figsize=(12, 6))  
    ax = pivot_df.plot(kind='line', color=[year_colors[year] for year in pivot_df.columns], marker='o')
    
    plt.xlabel('Distance to street façades', fontsize=12)
    plt.ylabel('Percentage (%)', fontsize=12)
    plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{int(x)}'))
    plt.xticks(fontsize=12)
    plt.yticks(fontsize=12)
    plt.ylim(0, 100)  # Set y-axis from 0 to 100
    plt.yticks(range(0, 101, 20), fontsize=12) 
    plt.legend(title='Year', loc='upper right', fontsize=12)
    
    save_figure("ED_Fig_1a_distance_line.svg")

# ED_Fig.1b. Distance to Streets Stacked Bar 
def create_EDfig1b(data):
    filtered_df = data[data['D_Bldback(m)'] != '<20']
    pivot_df = filtered_df.pivot_table(index='Year', columns='D_Bldback(m)', values='Percentage(%)', fill_value=0)
    order = ['20-40', '40-60', '60-80', '80-100', '100-120', '120-140', '>140']
    pivot_df = pivot_df[order]

    plt.figure(figsize=(20, 6))
    ax = pivot_df.plot(kind='bar', stacked=True, color=stacked_colors)
    
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Percentage (%)', fontsize=12)
    
    for p in ax.patches:
        height = p.get_height()
        width = p.get_width()
        x = p.get_x() + width / 2
        y = p.get_y() + height / 2
        ax.text(x, y, f'{height:.1f}', ha='center', va='center', fontsize=12, color='black')
    
    plt.xticks(rotation=0, fontsize=12)
    plt.yticks(fontsize=12)
    
    plt.legend(title='Distance to street façades (m)', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)
    plt.subplots_adjust(right=0.7)  
    
    save_figure("ED_Fig_1b_distance_stacked.svg")

# ED_Fig.1c. Layer Type Line Plot
def create_EDfig1c(data):   
    pivot_df = data.pivot_table(index='LayerType', columns='Year', values='Percentage(%)', fill_value=0)

    plt.figure(figsize=(10, 6)) 
    for year in [2015, 2019, 2023]:
        plt.plot(pivot_df.index, pivot_df[year], label=str(year), 
                color=year_colors[year], marker='o', linestyle='-', linewidth=2, markersize=6)
    
    plt.xlabel('Layer type', fontsize=12)
    plt.ylabel('Percentage (%)', fontsize=12)
    plt.xticks(rotation=0, fontsize=12)
    plt.yticks(fontsize=12)
    plt.ylim(0, 100)  # Set y-axis from 0 to 100
    plt.yticks(range(0, 101, 20), fontsize=12) 
    plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{int(x)}'))
    plt.legend(title='Year', loc='upper right', fontsize=12)
    
    save_figure("ED_Fig_1c_layer_line.svg")

# ED_Fig.1d. Layer Type Stacked Bar 
def create_EDfig1d(data):
    filtered_df = data[data['LayerType'] != '2 Ground floor']
    pivot_df = filtered_df.pivot_table(index='Year', columns='LayerType', values='Percentage(%)', fill_value=0)

    plt.figure(figsize=(20, 6))  
    ax = pivot_df.plot(kind='bar', stacked=True, color=stacked_colors)
    
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Percentage (%)', fontsize=12)
    
    for p in ax.patches:
        height = p.get_height()
        width = p.get_width()
        x = p.get_x() + width / 2
        y = p.get_y() + height / 2
        ax.text(x, y, f'{height:.2f}', ha='center', va='center', fontsize=12, color='black')
    
    plt.xticks(rotation=0, fontsize=12)
    plt.yticks(fontsize=12)
    
    plt.legend(title='Layer type', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)
    plt.subplots_adjust(right=0.7)  
    
    save_figure("ED_Fig_1d_layer_stacked.svg")

if __name__ == "__main__":  
    create_EDfig1a(Dback)   # ED_Fig.1a. Distance to Street Facade
    create_EDfig1b(Dback)   # ED_Fig.1b. Distance to Street Facade Stacked Bar 
    create_EDfig1c(LayerT)  # ED_Fig.1c. Layer Type 
    create_EDfig1d(LayerT)  # ED_Fig.1d. Layer Type Stacked Bar 
```

## ED_Fig. 2|Spatial distribution trends by establishment category (2015-2023)

```python
# Load data
DbackC = pd.read_csv(data_path+'RT_TypeD_Bldback.csv')
LayerTC = pd.read_csv(data_path+'RT_TypeLayertype.csv')

# Save figure
def save_figure(filename):
    plt.tight_layout()
    plt.savefig(f"{figure_path}/{filename}", format="svg", bbox_inches='tight')
    plt.show()

# Global settings
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 12
plt.rcParams['axes.titlesize'] = 12

# Color definitions
stacked_colors = ['#3851a1', '#5c90c2', '#92c6dc', '#ffe99d', '#feb76d', '#f67949', '#da382b']

# ED_Fig.2a-c NSF
def create_EDfig2abc(data):
    filtered_df = data[data['D_Bldback(m)'] != '<20'].copy()
    d_bldback_order = ['20-40', '40-60', '60-80', '80-100', '100-120', '120-140', '>140']
    filtered_df.loc[:, 'D_Bldback(m)'] = pd.Categorical(filtered_df['D_Bldback(m)'], 
                                                   categories=d_bldback_order, 
                                                   ordered=True)
    
    big_cate_order = ['F&B', 'Retail goods', 'Daily service', 'Exercise', 'Entertainment', 'Beauty', 'Training']
    filtered_df['Big_cate'] = pd.Categorical(filtered_df['Big_cate'], 
                                               categories=big_cate_order,
                                               ordered=True)
    
    for year in filtered_df['Year'].unique():
        year_data = filtered_df[filtered_df['Year'] == year].sort_values('Big_cate')
        pivot_df = year_data.pivot_table(index='Big_cate', 
                                       columns='D_Bldback(m)', 
                                       values='Percentage(%)', 
                                       fill_value=0)
        
        pivot_df = pivot_df[d_bldback_order]
        
        plt.figure(figsize=(10, 6))
        ax = pivot_df.plot(kind='bar', 
                          stacked=True, 
                          color=stacked_colors[:len(pivot_df.columns)],
                          width=0.8)
        
        plt.xlabel(year, fontsize=12)
        plt.ylabel('Percentage (%)', fontsize=12)
        plt.ylim(0, 70)
        plt.xticks(range(len(big_cate_order)), big_cate_order, fontsize=12, rotation=90, ha='right')
        plt.yticks(fontsize=12)
        
        plt.legend(title='Distance to streets (m)', 
                 bbox_to_anchor=(1.05, 1), 
                 loc='upper left', 
                 fontsize=12)
        save_figure(f"ED_Fig_2_distance_{year}.svg")

# ED_Fig.2d-f NGF
def create_EDfig2def(data):
    filtered_df = data[data['LayerType'] != '2 Ground floor'].copy()
    
    big_cate_order = ['F&B', 'Retail goods', 'Daily service', 'Exercise', 'Entertainment', 'Beauty', 'Training']
    filtered_df ['Big_cate'] = pd.Categorical(filtered_df ['Big_cate'],
                                               categories=big_cate_order,
                                               ordered=True)
    
    for year in filtered_df['Year'].unique():
        year_data = filtered_df[filtered_df['Year'] == year].sort_values('Big_cate')
        pivot_df = year_data.pivot_table(index='Big_cate',
                                       columns='LayerType',
                                       values='Percentage(%)',
                                       fill_value=0)
        
        plt.figure(figsize=(10, 6))
        ax = pivot_df.plot(kind='bar',
                          stacked=True,
                          color=stacked_colors[:len(pivot_df.columns)],
                          width=0.8)  
        
        plt.xlabel(year, fontsize=12)
        plt.ylabel('Percentage (%)', fontsize=12)
        plt.ylim(0, 40)
        plt.xticks(range(len(big_cate_order)), big_cate_order, fontsize=12, rotation=90, ha='right')
        plt.yticks(fontsize=12)
        plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{int(x)}'))
          
        plt.legend(title='Layer type',
                 bbox_to_anchor=(1.05, 1),
                 loc='upper left',
                 fontsize=12)
        save_figure(f"ED_Fig_2_layertype_{year}.svg")

# Main execution
if __name__ == "__main__":
    create_EDfig2abc(DbackC)    # ED_Fig.2a-c NSF
    create_EDfig2def(LayerTC)   # ED_Fig.2d-f NGF
```

## ED_Fig. 3|Spatial distribution trends by city socioeconomic tier (2015-2023)

```python
# Load data
DbackT = pd.read_csv(data_path+'RT_TierD_Bldback.csv')
LayerTT = pd.read_csv(data_path+'RT_TierLayertype.csv')

# Save figure
def save_figure(filename):
    plt.tight_layout()
    plt.savefig(f"{figure_path}/{filename}", format="svg", bbox_inches='tight')
    plt.show()
    
# Global settings
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 12
plt.rcParams['axes.titlesize'] = 12

# Color definitions
stacked_colors = ['#3851a1', '#5c90c2', '#92c6dc', '#ffe99d', '#feb76d', '#f67949', '#da382b']
citytier_order = ['Tier 1', 'Tier 2', 'Tier 3', 'Tier 4', 'Tier 5']

# ED_Fig.3a-c: NSF
def create_EDfig3abc(data):
    filtered_df = data[data['D_Bldback(m)'] != '<20'].copy()
    d_bldback_order = ['20-40', '40-60', '60-80', '80-100', '100-120', '120-140', '>140']
    filtered_df['D_Bldback(m)'] = pd.Categorical(filtered_df['D_Bldback(m)'], 
                                                    categories=d_bldback_order, 
                                                    ordered=True)
    
    for year in filtered_df['Year'].unique():
        year_data = filtered_df[filtered_df['Year'] == year]
        pivot_df = year_data.pivot_table(index='CityTier', 
                                       columns='D_Bldback(m)', 
                                       values='Percentage(%)', 
                                       fill_value=0)
        
        pivot_df = pivot_df.reindex(citytier_order)[d_bldback_order]
        
        plt.figure(figsize=(20, 6))
        pivot_df.plot(kind='bar', 
                     stacked=True, 
                     color=stacked_colors[:len(pivot_df.columns)],
                     width=0.8)
        
        plt.xlabel(year, fontsize=12)
        plt.ylabel('Percentage (%)', fontsize=12)
        plt.ylim(0, 55)
        plt.xticks(rotation=0, fontsize=12)
        plt.yticks(fontsize=12)
        
        plt.legend(title='Distance to streets (m)', 
                 bbox_to_anchor=(1.05, 1), 
                 loc='upper left', 
                  fontsize=12)
        save_figure(f"ED_Fig_3_distance_tier_{year}.svg")

# ED_Fig.3d-f: NGF
def create_EDfig3def(data):
    filtered_df = data[data['LayerType'] != '2 Ground floor'].copy()

    for year in filtered_df['Year'].unique():
        year_data = filtered_df[filtered_df['Year'] == year]
        pivot_df = year_data.pivot_table(index='CityTier', 
                                       columns='LayerType', 
                                       values='Percentage(%)', 
                                       fill_value=0)
        
        # Ensure correct ordering
        pivot_df = pivot_df.reindex(citytier_order)
        

        plt.figure(figsize=(20, 6))
        pivot_df.plot(kind='bar', 
                     stacked=True, 
                     color=stacked_colors[:len(pivot_df.columns)],
                     width=0.8)
        
        plt.xlabel(year, fontsize=12)
        plt.ylabel('Percentage (%)', fontsize=12)
        plt.ylim(0, 15)
        plt.xticks(rotation=0, fontsize=12)
        plt.yticks(fontsize=12)
        plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{int(x)}'))
        
        plt.legend(title='Layer type',
                 bbox_to_anchor=(1.05, 1), 
                 loc='upper left', 
                  fontsize=12)
        save_figure(f"ED_Fig_3_layertype_tier_{year}.svg")

# Main execution
if __name__ == "__main__":
    create_EDfig3abc(DbackT)    # ED_Fig.3a-c NSF
    create_EDfig3def(LayerTT)   # ED_Fig.3d-f NGF
```

## ED_Fig. 4|Results using different thresholds to identify non-street-facing shops

```python
# Load data
data = pd.read_csv(data_path+'RT_Thresholds.csv')

def save_figure(filename):
    plt.tight_layout()
    plt.savefig(f"{figure_path}/{filename}", format="svg", bbox_inches='tight')
    plt.show()
    

plt.rcParams.update({
    'font.family': 'Arial',
    'svg.fonttype': 'none',
    'figure.dpi': 300,
    'lines.markeredgewidth': 1.5,
    'lines.markersize': 10,
    'errorbar.capsize': 8
})

# ED_Fig.4: Threshold
def create_EDfig4(data):
    colors = {2015: '#6eaed1', 2019: '#fed06e', 2023: '#dc6e57'}
    
    g = sns.catplot(
        x='Threshold(m)', y='median_P_NSF(%)', hue='Year',
        data=data, kind='point', height=6, aspect=1.5,
        palette=list(colors.values()), 
        legend=False,
        errorbar=None
    )

    for year in [2015, 2019, 2023]:
        year_data = data[data['Year'] == year]
        for i, threshold in enumerate(sorted(year_data['Threshold(m)'].unique())):
            subset = year_data[year_data['Threshold(m)'] == threshold]
            median = subset['median_P_NSF(%)'].values[0]
            lower = subset['ci_lower95_P_NSF(%)'].values[0]
            upper = subset['ci_upper95_P_NSF(%)'].values[0]
            
            yerr = [[median - lower], [upper - median]]
            
            g.ax.errorbar(
                i, median, 
                yerr=yerr,
                fmt='o',
                color=colors[year], 
                linestyle='--',
                linewidth=2,
                markersize=5,
                capsize=10,
                capthick=2,
                elinewidth=2,
            )

    g.set_xticklabels(['10m', '15m', '20m', '25m', '30m'])
    g.set_axis_labels('Results using different thresholds', 
                     'Percentage of non-street-facing shops (%)')
    g.ax.set_ylim(0, 80)

    for spine in ['top', 'right']:
        g.ax.spines[spine].set_visible(True)

    g.ax.legend(title='Year', loc='upper right')

    g.fig.set_size_inches(8, 5)

    g.savefig(f"{figure_path}/ED_Fig_4_Threshold.svg", 
             bbox_inches='tight')
    plt.show()

# Main execution
create_EDfig4(data)  # ED_Fig.4 Threshold
```

## ED_Fig. 5|Results using different building datasets

```python
# Load data
Data = pd.read_csv(data_path+'RT_NSFMainstudy120.csv')
DataC = pd.read_csv(data_path+'RT_CityNSFMainstudy120.csv')
DataRT = pd.read_csv(data_path+'RT_NSFRobusttest120.csv')
DataRTC = pd.read_csv(data_path+'RT_CityNSFRobusttest120.csv')

# Save figure 
def save_figure(filename):
    plt.tight_layout()
    plt.savefig(f"{figure_path}/{filename}", format="svg", bbox_inches='tight')
    plt.show()

# Global settings
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 12
plt.rcParams['axes.titlesize'] = 12

# Color definitions
year_colors = {2015: '#6eaed1', 2019: '#fed06e', 2023: '#dc6e57'}


# ED_Fig.5a - Dual Axis Plot for Data
def create_EDfig5a(data):  
    plt.figure(figsize=(4, 4))
    
    bars = plt.bar(data['Year'], data['NSF(millions)'], 
                  color=[year_colors[year] for year in data['Year']], width=0.9)

    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2, height, f'{height:.2f}', 
                ha='center', va='bottom', fontsize=10)

    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Count of non-street-facing shops (millions)', fontsize=10)
    plt.xticks(data['Year'], fontsize=10)
    plt.ylim(0, 10)
    plt.yticks(fontsize=10)

    ax2 = plt.gca().twinx()
    ax2.plot(data['Year'], data['P_NSF(%)'], color='red', marker='o', 
            label='Percentage of non-street-facing shops (%)')
    ax2.set_ylabel('Percentage of non-street-facing shops (%)', fontsize=10, color='red')
    ax2.tick_params(axis='y', labelcolor='red', labelsize=10)
    ax2.set_ylim(0, 51)

    offset = 50 * 0.03
    for x, y in zip(data['Year'], data['P_NSF(%)']):
        ax2.text(x, y - offset, f'{y:.2f}', ha='center', va='top', fontsize=10, color='red')
    
    save_figure("ED_Fig_5a_dual_axis.svg")

# ED_Fig.5b - KDE Plot for Data
def create_EDfig5b(data): 
    fig, ax = plt.subplots(figsize=(8, 5), dpi=300)
    colors = {2015: '#6eaed1', 2019: '#fed06e', 2023: '#dc6e57'}
    
    sns.kdeplot(data=data, x='P_NSF(%)', hue='Year',
               fill=True, bw_adjust=0.15, palette=colors,
               hue_order=[2015, 2019, 2023])
    
    ymin, ymax = ax.get_ylim()
    label_y = 0.051
    
    legend_handles = []
    
    for year in [2015, 2019, 2023]:
        mean_val = data[data['Year'] == year]['P_NSF(%)'].mean()
        color = colors[year]
        
        line = plt.axvline(mean_val, color=color, linestyle='dashed',
                          linewidth=1, label=f'Mean ({year})')
        legend_handles.append(line)
        
        if year == 2015:
            label_x = mean_val - 1.4
        else:
            label_x = mean_val + 1.4
            
        plt.text(label_x, label_y, f'{mean_val:.2f}',
                color=color, ha='center', va='top', fontsize=10,
                bbox=dict(facecolor='white', alpha=0.8, edgecolor='none', pad=1))
    
    plt.xlabel('Percentage of non-ground-floor shops (%)', fontsize=12)
    plt.ylabel('Probability density', fontsize=12)
    plt.ylim(0, 0.052)
    plt.legend(handles=legend_handles, fontsize=10, title='Year', title_fontsize=11)
    
    
    save_figure("ED_Fig_5b_kde.svg")

# ED_Fig.5c - Dual Axis Plot for DataRT
def create_EDfig5c(data):
    
    plt.figure(figsize=(4, 4))

    bars = plt.bar(data['Year'], data['NSF(millions)'], 
                  color=[year_colors[year] for year in data['Year']], width=0.9)

    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2, height, f'{height:.2f}', 
                ha='center', va='bottom', fontsize=10)

    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Count of non-street-facing shops (millions)', fontsize=10)
    plt.xticks(data['Year'], fontsize=10)
    plt.ylim(0, 10)
    plt.yticks(fontsize=10)

    ax2 = plt.gca().twinx()
    ax2.plot(data['Year'], data['P_NSF(%)'], color='red', marker='o', 
            label='Percentage of non-street-facing shops (%)')
    ax2.set_ylabel('Percentage of non-street-facing shops (%)', fontsize=10, color='red')
    ax2.tick_params(axis='y', labelcolor='red', labelsize=10)
    ax2.set_ylim(0, 51)

    offset = 50 * 0.03
    for x, y in zip(data['Year'], data['P_NSF(%)']):
        ax2.text(x, y - offset, f'{y:.2f}', ha='center', va='top', fontsize=10, color='red')
    
    save_figure("ED_Fig_5c_dual_axis_RT.svg")

# ED_Fig.5d - KDE Plot for DataRT
def create_EDfig5d(data):

    fig, ax = plt.subplots(figsize=(8, 5), dpi=300)
    colors = {2015: '#6eaed1', 2019: '#fed06e', 2023: '#dc6e57'}
    
    sns.kdeplot(data=data, x='P_NSF(%)', hue='Year',
               fill=True, bw_adjust=0.15, palette=colors,
               hue_order=[2015, 2019, 2023])
    
    ymin, ymax = ax.get_ylim()
    label_y = 0.051
    
    legend_handles = []
    
    for year in [2015, 2019, 2023]:
        mean_val = data[data['Year'] == year]['P_NSF(%)'].mean()
        color = colors[year]
        
        line = plt.axvline(mean_val, color=color, linestyle='dashed',
                          linewidth=1, label=f'Mean ({year})')
        legend_handles.append(line)
        
        if year == 2015:
            label_x = mean_val - 1.6
        else:
            label_x = mean_val + 1.6
            
        plt.text(label_x, label_y, f'{mean_val:.2f}',
                color=color, ha='center', va='top', fontsize=10,
                bbox=dict(facecolor='white', alpha=0.8, edgecolor='none', pad=1))
    
    plt.xlabel('Percentage of non-ground-floor shops (%)', fontsize=12)
    plt.ylabel('Probability density', fontsize=12)
    
    plt.legend(handles=legend_handles, fontsize=10, title='Year', title_fontsize=11)
    plt.ylim(0, 0.052)
    save_figure("ED_Fig_5d_kde_RT.svg")


if __name__ == "__main__":
    create_EDfig5a(Data)     # ED_Fig.5a - Dual Axis Plot for Data
    create_EDfig5b(DataC)    # ED_Fig.5b - KDE Plot for Data
    create_EDfig5c(DataRT)   # ED_Fig.5c - Dual Axis Plot for DataRT
    create_EDfig5d(DataRTC)  # ED_Fig.5d - KDE Plot for DataRT
```
